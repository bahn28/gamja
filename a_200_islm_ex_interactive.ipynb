```python
# %% [markdown]
# ## 마우스로 조절하는 IS-LM 모형 시뮬레이션

import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider, IntSlider # [1]
```

```python

# 1. 고정된 기초 경제 파라미터 설정
c0, c1 = 200, 0.75  # 기초소비, 한계소비성향
i0, i1 = 300, 1000  # 기초투자, 투자의 이자율 민감도
T = 200            # 세금
l1, l2 = 0.5, 1500  # 화폐수요의 소득/이자율 민감도

# X축 (국민소득 Y의 범위)
Y = np.linspace(1200, 3000, 1000)
```

```python

# 2. 슬라이더 조작 시 실행될 그래프 업데이트 함수 정의
def update_islm(G=400, M_P=600):
    """
    G: 정부지출 (재정정책 변수)
    M_P: 실질화폐공급량 (통화정책 변수)
    """
    # IS, LM 곡선 수식 (r에 대해 정리)
    r_IS = (c0 - c1*T + i0 + G - (1 - c1)*Y) / i1
    r_LM = (l1*Y - M_P) / l2
    
    # 두 곡선의 교점(균형 국민소득과 균형 이자율) 수학적 계산
    # 연립방정식을 풀어 Y_eq와 r_eq를 도출
    Y_eq = (c0 - c1*T + i0 + G + (i1/l2)*M_P) / ((1 - c1) + (i1/l2)*l1)
    r_eq = (l1*Y_eq - M_P) / l2
    
    # 그래프 그리기 시작
    plt.figure(figsize=(10, 6))
    plt.plot(Y, r_IS, label='IS Curve (Goods Market)', color='#1f77b4', linewidth=2.5)
    plt.plot(Y, r_LM, label='LM Curve (Money Market)', color='#ff7f0e', linewidth=2.5)
    
    # 균형점(교점) 시각화
    plt.scatter(Y_eq, r_eq, color='red', s=120, zorder=5, 
                label=f'Equilibrium (Y={Y_eq:.0f}, r={r_eq*100:.1f}%)')
    
    # 균형점 점선 안내선
    plt.axvline(Y_eq, color='gray', linestyle='--', alpha=0.7)
    plt.axhline(r_eq, color='gray', linestyle='--', alpha=0.7)
    
    # 그래프 스타일 및 축 고정 (축이 고정되어야 곡선의 이동이 잘 보입니다)
    plt.title('Interactive IS-LM Model Simulation', fontsize=14, fontweight='bold', pad=15)
    plt.xlabel('National Income (Y)', fontsize=12)
    plt.ylabel('Interest Rate (r)', fontsize=12)
    plt.xlim(1200, 3000)
    plt.ylim(-0.05, 0.55) # 이자율 -5% ~ 55% 범위 고정
    plt.grid(True, linestyle=':', alpha=0.6)
    plt.legend(loc='upper right', fontsize=11)
    plt.show()
```

```python

# 3. 대화형 위젯 연결 (정부지출 G와 화폐공급 M_P 슬라이더 생성)
interact(
    update_islm, 
    G=IntSlider(min=200, max=700, step=50, value=400, description='정부지출(G):'),
    M_P=IntSlider(min=300, max=900, step=50, value=600, description='통화량(M/P):')
); # [1]
```
