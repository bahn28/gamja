```python
# %% [markdown]
# ## 유동성 선호 모형으로부터 LM 곡선 유도 (수평선 정렬 완벽 보정)

import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, IntSlider
```

Money Market. 실질잔고의 공급, 실질잔고에 대한 수요
\begin{align}
(M/P)^s & = { M_0 \over P_0 } \\
(M/P)^d & = l_1 \times Y - l_2 \times r 
\end{align}

$ Y $와 $ r $의 관계는 어떤 모습일까?

```python

# 1. 고정 경제 파라미터 및 정책 변수
P = 1.0            # 단기 고정 물가 수준
l1 = 0.50          # 화폐수요의 소득 민감도
l2 = 1500          # 화폐수요의 이자율 민감도
M = 600            # 고정된 명목 통화량

# 좌측 그래프용 X축: 실질 화폐 잔고 (M/P)의 범위
MD_range = np.linspace(200, 1000, 1000)

# 우측 그래프용 X축: 국민소득 (Y)의 범위
Y_vals = np.linspace(1200, 2800, 1000)
```

```python

def derive_lm_horizontal_fixed(Y=2000):
    """
    Y: 현재 국민소득 (슬라이더로 조절하여 화폐수요 시프트 유도)
    """
    # --- [계산부] ---
    MS_real = M / P  # 실질 화폐 공급 (수직선 위치)
    
    # 좌측 그래프용 이자율 범위에 따른 화폐수요량 역산
    r_upper = np.linspace(0.01, 0.42, 100)
    MD_vals = l1 * Y - l2 * r_upper
    
    # 현재 소득(Y) 하에서의 균형 이자율(r_eq) 계산
    r_eq = (l1 * Y - MS_real) / l2
    
    # 전체 LM 곡선 데이터
    r_LM = (l1 * Y_vals - MS_real) / l2
    
    # --- [시각화부: 좌우 배치] ---
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))
    
    # [좌측 그래프: 유동성 선호 이론 (화폐 시장)]
    ax1.axvline(MS_real, label=f'Money Supply (M/P = {MS_real:.0f})', color='orange', linewidth=2.5)
    ax1.plot(MD_vals, r_upper, label=f'Money Demand L(Y={Y}, r)', color='#1f77b4', linewidth=2.5)
    ax1.scatter(MS_real, r_eq, color='red', s=100, zorder=5, label=f'Equilibrium r* = {r_eq*100:.1f}%')
    
    # [보정 핵심] 두 차트에 각각 독립적인 수평 가이드 점선을 추가하여 왜곡 방지
    ax1.axhline(r_eq, color='red', linestyle='--', alpha=0.5, linewidth=1.5)
    
    ax1.set_title('1. Theory of Liquidity Preference', fontsize=12, fontweight='bold')
    ax1.set_xlabel('Real Money Balances (M/P)')
    ax1.set_ylabel('Interest Rate (r)')
    ax1.set_xlim(200, 1000)
    ax1.set_ylim(0.00, 0.42)
    ax1.grid(True, linestyle=':', alpha=0.5)
    ax1.legend(loc='upper right')
    
    # [우측 그래프: LM 곡선 평면]
    ax2.plot(Y_vals, r_LM, label='LM Curve', color='purple', linewidth=2.5)
    ax2.scatter(Y, r_eq, color='red', marker='o', s=120, zorder=5, label='Current Equilibrium Point')
    
    # [보정 핵심] 우측 차트에도 동일한 높이의 수평 가이드 점선과 소득 가이드 수직선 추가
    ax2.axhline(r_eq, color='red', linestyle='--', alpha=0.5, linewidth=1.5)
    ax2.axvline(Y, color='gray', linestyle=':', alpha=0.5)
    
    ax2.set_title('2. LM Curve Derivation (Y-r Space)', fontsize=12, fontweight='bold')
    ax2.set_xlabel('National Income / Output (Y)')
    ax2.set_ylabel('Interest Rate (r)')
    ax2.set_xlim(1200, 2800)
    ax2.set_ylim(0.00, 0.42)
    ax2.grid(True, linestyle=':', alpha=0.5)
    ax2.legend(loc='upper left')
    
    plt.tight_layout()
    plt.show()
```

```python

# 3. 국민소득(Y)을 조절하는 슬라이더 위젯 생성
interact(
    derive_lm_horizontal_fixed,
    Y=IntSlider(min=1400, max=2600, step=100, value=2000, description='국민소득(Y):')
);
```
