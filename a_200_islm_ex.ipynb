```python
import numpy as np
import matplotlib.pyplot as plt
```

GM2, MM2,
\begin{align}
PE & = Y \\
PE & = c_0 + c_1(Y-T) + I + G \\
(M/P)^s & = { M_0 \over P_0 } \\
L(Y,r)  & = l_1 \times Y - l_2 \times r 
\end{align}

```python

# 1. 경제 파라미터 및 정책 변수 설정
c0, c1 = 200, 0.75  # 기초소비, 한계소비성향
i0, i1 = 300, 1000  # 기초투자, 투자의 이자율 민감도
G = 400            # 정부지출 (정책 변수 1)
T = 200            # 세금

M_P = 600          # 실질 화폐 공급량 (정책 변수 2)
l1, l2 = 0.5, 1500  # 화폐수요의 소득 민감도, 이자율 민감도
```

\begin{align}
\text{IS: } r & = { (c_0 - c_1 \times T + i_0 + G_0 - (1 - c_1)*Y) \over i_1 } \\
\text{LM: } r & =  { (l_1 \times Y - M_0/P_0) \over l_2 } \\
\text{EQ: }Y^* &= { (c_0 - c_1\times T + i_0 + G_0 + (i_1/l_2) \times M_0/P_0) 
                        \over  ((1 - c_1) + (i_1/l_2\times l_1)) } \\
\text{EQ: }r^* & = { (l_1\times Y^* - M_0/P_0) \over l_2 }
\end{align}

```python

# 2. 소득(Y) 범위 설정 (그래프의 X축)
Y = np.linspace(1500, 2800, 500)

# 3. IS 곡선 공식 정의 (r에 대해 정리)
# Y = c0 + c1*(Y - T) + i0 - i1*r + G
# r = (c0 - c1*T + i0 + G - (1 - c1)*Y) / i1
r_IS = (c0 - c1*T + i0 + G - (1 - c1)*Y) / i1

# 4. LM 곡선 공식 정의 (r에 대해 정리)
# M_P = l1*Y - l2*r
# r = (l1*Y - M_P) / l2
r_LM = (l1*Y - M_P) / l2

# 연립방정식을 풀어 Y_eq와 r_eq를 도출
Y_eq = (c0 - c1*T + i0 + G + (i1/l2)*M_P) / ((1 - c1) + (i1/l2)*l1)
r_eq = (l1*Y_eq - M_P) / l2
```

```python

# 6. 그래프 그리기
plt.figure(figsize=(10, 6))
plt.plot(Y, r_IS, label='IS Curve (Goods Market)', color='blue', linewidth=2)
plt.plot(Y, r_LM, label='LM Curve (Money Market)', color='orange', linewidth=2)

# 균형점 표시
plt.scatter(Y_eq, r_eq, color='red', s=100, zorder=5, label=f'Equilibrium (Y={Y_eq:.0f}, r={r_eq*100:.1f}%)')
plt.axvline(Y_eq, color='gray', linestyle='--')
plt.axhline(r_eq, color='gray', linestyle='--')

# 그래프 스타일링
plt.title('IS-LM Model Simulation using Python', fontsize=14, fontweight='bold')
plt.xlabel('National Income (Y)', fontsize=12)
plt.ylabel('Interest Rate (r)', fontsize=12)
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend(fontsize=11)
plt.show()
```
