# 01. 선형 회귀(Linear Regression) 소개

이 페이지에서는 머신러닝의 가장 기초적인 모델인 **선형 회귀**의 수학적 배경을 알아보고, 
파이썬 코드로 실제 데이터를 시각화해 봅니다.

## 1. 수학적 모델 정의

선형 회귀 모델은 독립 변수 $x$와 종속 변수 $y$의 관계를 다음과 같은 선형 방정식으로 가정한 모델입니다.

$$ y = w \cdot x + b $$

여기서 각 기호의 의미는 다음과 같습니다:
- $w$ (Weight): 회귀 계수 (기울기)
- $b$ (Bias): 편향 (y절편)

실제 데이터와 모델의 예측값 사이의 차이를 계산하기 위해 **평균 제곱 오차(MSE, Mean Squared Error)** 손실 함수를 사용합니다.

$$ L(w, b) = \frac{1}{n} \sum_{i=1}^{n} \left( y_i - (w \cdot x_i + b) \right)^2 $$


```python
import numpy as np
import matplotlib.pyplot as plt

# 랜덤 시드 고정
np.random.seed(42)

# 샘플 데이터 생성 (가상의 선형 데이터 + 노이즈)
X = 2 * np.random.rand(100, 1)
y = 4 + 3 * X + np.random.randn(100, 1)

print(f"데이터 생성 완료! 총 {len(X)}개의 샘플이 준비되었습니다.")
```

## 2. 데이터 시각화

생성된 데이터의 분포를 `matplotlib` 라이브러리를 사용하여 산점도(Scatter plot)로 시각화해 보겠습니다.


```python
plt.figure(figsize=(8, 5))
plt.scatter(X, y, color='blue', alpha=0.6, label='Actual Data')
plt.title('Linear Regression Sample Data')
plt.xlabel('X (Independent Variable)')
plt.ylabel('y (Dependent Variable)')
plt.grid(True)
plt.legend()
plt.show()
```
